```python
# Model A — unified sonar object detector
# Merges Sub_pipe (Pipeline), SeabedObjects-KLSG (Aircraft/Fish/Other/Shipwreck),
# mine detection (MILCO/NOMBO), and marine-debris watertank-segmentation
# (Tire/Bottle/Drink-carton/Chain/Can/Valve/Propeller/Hook/Shampoo-bottle/Standing-bottle)
# into one 17-class YOLO detector. See ~/SIH 2026/sonar-debris/model_a_unified/data.yaml for the merge.

from ultralytics import YOLO
import os

RUNS_DIR = os.path.expanduser("~/SIH 2026/sonar-debris/runs/detect")
DATA_YAML = os.path.expanduser("~/SIH 2026/sonar-debris/model_a_unified/data.yaml")

model = YOLO("yolov8s.pt")

results = model.train(
    data=DATA_YAML,
    epochs=60,
    imgsz=640,
    batch=24,

    # Sonar has no meaningful hue; keep some brightness/scale jitter for the mixed sensors
    hsv_h=0.0,
    hsv_s=0.0,
    hsv_v=0.3,
    mosaic=0.7,
    mixup=0.0,
    degrees=10.0,
    translate=0.1,
    scale=0.4,
    shear=2.0,

    patience=15,

    project=RUNS_DIR,
    name="model_a_unified",
    exist_ok=True,

    device=0,
)
```
